````python
def student_max_marks(marks):

  highest_marks = list()
  current_highest = 0

  for i in range(len(marks)):
    for j in range(len(marks[0])):
      if marks[i][j] > current_highest:# To check the current highest marks of student
        current_highest = marks[i][j] # storing highest marks till now in variable current_highest

    highest_marks.append(current_highest)# appending highest marks of student i at ith index of highest_marks
    current_highest = 0 # re-setting the value before checking next student

  for i in range(len(marks)):
    print("student ",i + 1, " max : ", highest_marks[i] )


  return highest_marks

````

Define two functions:

filter_evens(numbers, make_negative=False) that:


Takes a list of numbers as first argument


Takes a default parameter make_negative=False


Uses a list comprehension and is_even() inside it:


If make_negative is False: returns a list of only the even numbers


If make_negative is True: returns a list of even numbers but multiplied by -1


Example:
 filter_evens([1, 2, 3, 4], make_negative=True) → [-2, -4].


````python
def is_even(n):
  if n % 2 == 0:
    return True
  else:
    return False

def filter_evens(numbers, make_negative = False):
  if make_negative:
    return[-n for n in numbers if is_even(n) ]
  else:
    return[n for n in numbers if is_even(n)]



````

For immutable data types, when the function modifies the parameter, it actually creates a new object within the function's local scope, and the original object outside the function remains unchanged .
Common immutable types include:

For mutable data types, both the function parameter and the original variable outside the function point to the same object in memory. If the function modifies the contents of the object in place, these changes will be reflected outside the function .

## Recursion


Definition :- Recursion is defined as a processes which call itself directly or indirectly and the corresponding function is called recursive.

Example : suppose u r standing on a staircase with n steps. You want to write a function that climbs down the stairs. You will write a function that repeatedly climbs down the stairs one step at a time until you land at the last step i.e n = 0.

A Recursion Function can be broken in 3 parts :


1.   Base Case : This is where recursive function stops. In the above example When we reach at last step n = 0 is our base case.
2.   Work : This is the broken down task that the function repeatedly does. In the above example getting down one step is the work.
3.   Recursive call : This is the call that keeps the reaction going until all the steps are completed. A function normally would terminate after doing its work, So the recursive call is the call to the same fuction inside the fuction




## Recursive Function Example :-

````python
def countdown(n):
  if n == 0: # Base Case
    print("Blast off")
    return
  #print(n) # Work
  countdown(n-1) # Recursive Call
  print(n)
````

````python
countdown(5)
````
